## 3) Типы данных и `type()`

### Тип — это “вид” значения
- `int` — целое число (10, -3)
- `float` — дробное число (2.5, 3.14)
- `str` — строка ("текст")

### `type(x)`
Показывает тип значения в переменной `x`.

### Зачем
Чтобы понимать, что вы храните и какие операции допустимы.


````
x = 10
y = 2.5
name = "AI"

print(type(x), type(y), type(name))
````

## 4) Функции — “ваши собственные команды”

### Что такое функция
Функция — это именованный кусочек кода, который можно вызывать много раз.

### Почему в курсе всё через функции
- удобно переносить в проект (VS Code)
- удобно проверять через `assert`
- код становится аккуратным

### Синтаксис
```python
def greet(name):
    return "Привет, " + name
```
- `def` — определить функцию
- `greet` — имя функции
- `(name)` — параметр (входные данные)
- `return` — вернуть результат


````
def greet(name: str) -> str:
    # Получаем имя и возвращаем строку приветствия
    return f"Привет, {name}!"

print(greet("Иван"))
````

## 5) Списки `list` — набор значений по порядку

### Что такое список
Список хранит несколько значений в порядке добавления: `[1, 2, 3]`

### Полезные функции
- `len(list)` — сколько элементов
- `sum(list)` — сумма чисел в списке

### Зачем
Так мы будем хранить коллекцию данных (например, items в дипломе).


````
nums = [1, 2, 3, 4, 5]
print("Список:", nums)
print("Сколько элементов:", len(nums))
print("Сумма:", sum(nums))
````

## 6) Кортежи `tuple` — как список, но “нельзя менять”

### Что такое кортеж
Кортеж: `(10, 20)` — хранит несколько значений, но **неизменяем**.

### Зачем
Удобно возвращать из функции несколько результатов сразу.


````
def arithmetic_ops(a: int, b: int):
    # Возвращаем 3 результата кортежем (sum, diff, prod)
    return (a + b, a - b, a * b)

print(arithmetic_ops(7, 3))
````

## 7) Словари `dict` — “ключ → значение”

### Что такое словарь
Словарь хранит значения по ключам (как анкета/карточка):
```python
student = {"name": "Ann", "group": "G1"}
```

### Доступ к данным
- `d["key"]` — прямой доступ (если ключа нет → ошибка `KeyError`)
- `d.get("key", default)` — безопасный доступ (если ключа нет → default)

### Зачем
Словарь подходит для “паспорта проекта”: title/goal/user/risks.


````
student = {"name": "Ann", "group": "G1"}

print("Имя (прямой доступ):", student["name"])
print("Группа (безопасно):", student.get("group", "UNKNOWN"))
print("Если ключа нет:", student.get("email", "нет email"))
````

## 9) Задача урока: паспорт проекта (первый артефакт диплома)

Сделаем функцию `project_passport(...)`, которая возвращает словарь с полями:
- `title` — название проекта
- `goal` — цель
- `user` — для кого проект
- `risks` — список рисков


````
def project_passport(title: str, goal: str, user: str, risks: list[str]) -> dict:
    return {"title": title, "goal": goal, "user": user, "risks": risks}

pp = project_passport(
    title="AI Helper",
    goal="Помогать пользователю",
    user="HR",
    risks=["data quality", "privacy"],
)

print("Паспорт проекта (как словарь):")
print(pp)
````

---
## 10) Красивый вывод: “печать паспорта проекта”

Сделаем функцию, которая превращает словарь паспорта в 4 строки текста.
Это удобно для консоли, отчётов и логов.


````
def pretty_passport_lines(passport: dict) -> list[str]:
    title = passport.get("title", "")
    goal = passport.get("goal", "")
    user = passport.get("user", "")
    risks = passport.get("risks", [])
    return [
        f"TITLE: {title}",
        f"GOAL: {goal}",
        f"USER: {user}",
        f"RISKS: {', '.join(risks)}",
    ]

print("\n".join(pretty_passport_lines(pp)))
````

## 11) Автопроверка: `assert`

### Что такое assert
`assert условие` — проверка. Если условие ложное, выполнение останавливается и показывает ошибку.

### Зачем
Это “встроенный преподаватель”: сразу видно, правильно ли написан код.

Ниже — проверки для всего занятия. Их НЕ меняем.


````
# =========================
# LESSON 01 — FULL TESTS (НЕ МЕНЯТЬ)
# =========================

def hello_lines(name: str) -> list[str]:
    return ["Hello, AI Diploma!", name]

def course_line(course_name: str) -> str:
    return f"Course: {course_name}"

def types_of(x, y, name):
    return (type(x), type(y), type(name))

def string_len(s: str) -> int:
    return len(s)

def sum_list(nums: list[int]) -> int:
    return sum(nums)

def student_fields(student: dict) -> tuple:
    return (student.get("name", ""), student.get("group", ""))

def run_all_tests():
    assert hello_lines("Ivan") == ["Hello, AI Diploma!", "Ivan"]
    assert course_line("grAI621") == "Course: grAI621"
    assert arithmetic_ops(7, 3) == (10, 4, 21)

    t = types_of(10, 2.5, "AI")
    assert t[0] is int and t[1] is float and t[2] is str

    assert greet("Иван") == "Привет, Иван!"
    assert string_len("Python") == 6
    assert string_len("") == 0

    assert sum_list([1, 2, 3, 4, 5]) == 15
    assert sum_list([]) == 0

    assert student_fields({"name": "Ann", "group": "G1"}) == ("Ann", "G1")

    pp2 = project_passport("AI Helper", "Помогать", "HR", ["data quality", "privacy"])
    assert pp2["title"] == "AI Helper"
    assert pp2["risks"] == ["data quality", "privacy"]

    lines = pretty_passport_lines(pp2)
    assert len(lines) == 4
    assert lines[0].startswith("TITLE:")
    assert "RISKS:" in lines[3]

    print("✅ LESSON 01: all tests passed")

run_all_tests()
````
